English OCR with contour-based localization. First use downloads pretrained OCR weights. The sample labels are historical, not benchmark results.

```python
from google.colab import files
import easyocr
import matplotlib.pyplot as plt
import cv2
from anpr.pipeline import read_image, recognize

reader = easyocr.Reader(["en"], gpu=False)
uploaded = files.upload()
```

```python
for filename in uploaded:
    result, stages = recognize(read_image(filename), reader)
    print(filename, result)
    for title, image in stages.items():
        plt.figure(figsize=(10, 5))
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB) if image.ndim == 3 else image, cmap="gray")
        plt.title(title)
        plt.axis("off")
        plt.show()
```
